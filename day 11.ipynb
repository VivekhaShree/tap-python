```python
import csv
from pathlib import Path
DATA_DIR = Path(r"C:\ProgramData\MySQL\MySQL Server 8.0\Uploads")
ORDERS_FILE = DATA_DIR / "olist_orders_dataset.csv"
print("Data folder:", DATA_DIR.resolve())
print("Orders file:", ORDERS_FILE)
```

```python
with ORDERS_FILE.open("r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)
    orders = list(reader)
print(len(orders))
```

```python
print("Column names:")
print(reader.fieldnames)
print("\nFirst 5 rows:")
for row in orders[:5]:
    print(row)
```

```python
missing_counts = {column: 0 for column in reader.fieldnames}
for row in orders:
    for column in reader.fieldnames:
        if row[column] is None or row[column].strip() == "":
            missing_counts[column] += 1
print("\nMissing values:")
for column, count in missing_counts.items():
    print(f"{column}: {count}")
```

```python
csv_files = sorted(DATA_DIR.glob("*.csv"))
for csv_file in csv_files:
    with csv_file.open("r", encoding="utf-8-sig", newline="") as file:
        csv_reader = csv.DictReader(file)
        columns = csv_reader.fieldnames or []
        count = 0
        missing = {c: 0 for c in columns}
        for row in csv_reader:
            count += 1
            for c in columns:
                if row[c] is None or row[c].strip() == "":
                    missing[c] += 1
    print(f"\nFile: {csv_file.name}")
    print("Rows:", count)
    print("Columns:", columns)
    print("Missing values:", missing)
```

# Reading CSV Files Using Python

This notebook demonstrates how to read CSV files using Pandas and parameterize file paths for reusable data loading.

```python

# Parameter Cell

from pathlib import Path

INPUT_FOLDER = Path(
    r"C:\ProgramData\MySQL\MySQL Server 8.0\Uploads"
)

FILE_NAME = "olist_orders_dataset.csv"

ENCODING = "utf-8-sig"
```

```python
import pandas as pd
```

```python
file_path = INPUT_FOLDER / FILE_NAME

df = pd.read_csv(
    file_path,
    encoding=ENCODING
)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```
